# Bayesian Modeling Guidelines  


---

### 1. **Understand the Data-Generating Process**
- Model the **process**, not just patterns in the data.
- Use **causal reasoning** and **DAGs** (Directed Acyclic Graphs).
- Avoid models that fit data but don’t reflect how the world works.

---

### 2. **Model the Uncertainty, Not Just the Mean**
- Include **full probability models**, not just point estimates.
- Model variability (e.g. σ) along with the mean.
- Use flexible likelihoods when appropriate (e.g. Negative Binomial, varying σ).

---

### 3. **Use Thoughtful, Informative Priors**
- Priors are part of the model — not optional.
- Use **prior predictive simulation** to ensure priors make sense.
- Prefer **weakly informative** priors over flat ones.

---

### 4. **Simulate Before You Estimate**
- Simulate data from your model before fitting it.
- Ask: Can this model generate plausible data?
- Simulation helps catch mis-specification early.

---

### 5. **Compare Models Predictively**
- Use **WAIC**, **LOO**, or **Bayes Factors** for model comparison.
- Focus on **predictive accuracy**, not just in-sample fit.
- Prefer simpler models if they generalize better.

---

### 6. **Model Multilevel (Hierarchical) Structure**
- Use **partial pooling** for nested or grouped data.
- Include **varying intercepts/slopes** as needed.
- This improves estimates and handles imbalance in data.

---

### 7. **Think Causally — Control for Confounding Intelligently**
- Use **DAGs** to decide what to condition on.
- Avoid adjusting for **colliders** or post-treatment variables.
- Apply the **backdoor criterion** to select adjustment sets.

---

### 8. **Perform Posterior Predictive Checks**
- Simulate from the **posterior predictive distribution**.
- Compare simulated vs observed data.
- Use checks to detect model misfit and refine structure.

---

### 9. **Acknowledge Model Limitations**
- All models are simplifications — be transparent about what’s missing.
- Communicate **uncertainty** honestly.
- Don’t oversell precision or conclusions.

---

### 10. **Build Models Incrementally**
- Start **simple**, add complexity only as needed.
- Validate with diagnostics and predictive checks at each step.
- Avoid unnecessary complexity that doesn’t improve understanding or prediction.

---

> **“Bayesian statistics is not about belief. It's about modeling uncertainty given assumptions.”**  
> — Richard McElreath


### Simple 4-step plan for honest digital scholarship (lecture 11)

1. **What are we trying to describe?**

2. **What is the ideal data for doing so?**

3. **What data do we actually have?**

4. **What causes the differences between (2) and (3)?**

5. *[optional]* Is there a way to use (3) + (4) to do (1)?
