You're on the right track with the model structure, but let me help clarify the likelihood difference.

The key insight is about **combinatorial terms**:

- **Disaggregated**: Each observation is a single Bernoulli trial (0 or 1), so the likelihood is just the product of individual probabilities
- **Aggregated**: When you have n trials with k successes, the likelihood includes the binomial coefficient C(n,k) = n!/(k!(n-k)!)

The regression coefficients and fitted probabilities are the same, but the actual likelihood values differ because the aggregated form includes these combinatorial terms that count the number of ways to get k successes in n trials.

**AI Prompt**

**AI Response**

Exactly right! You've identified the key reason.

The logit function logit(p) = log(p/(1-p)) does indeed map probabilities from the constrained range (0,1) to the unrestricted range (-∞, +∞). This allows the linear predictor α + βx to take any real value while ensuring the predicted probabilities always stay between 0 and 1.

There's also another nice property: the logit is the natural parameter of the binomial distribution in the exponential family, which makes it the "canonical" link function.
